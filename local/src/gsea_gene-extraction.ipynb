```python
def transpose_gene_symbols(input_file, output_file):
    with open(input_file, 'r') as f:
        for line in f:
            if line.startswith('GENE_SYMBOLS'):
                genes = line.split('\t')[1].strip()
                gene_list = genes.split(',')
                
                with open(output_file, 'w') as out:
                    for gene in gene_list:
                        out.write(f"{gene}\n")
                break
```

```python
input_file = "/mnt/trcanmed/snaketree/prj/DE_RNASeq/local/share/data/HALLMARK_KRAS_SIGNALING_UP.v2024.1.Hs.tsv"
output_file = "/mnt/trcanmed/snaketree/prj/DE_RNASeq/local/share/data/HALLMARK_KRAS_SIGNALING_UP_genes.txt"
```
